```python
from datasets import Dataset
import pandas as pd
import os

os.makedirs("input", exist_ok=True)

train = Dataset.from_file("banking77-train.arrow")
test = Dataset.from_file("banking77-test.arrow")

train.to_parquet("input/train.parquet")
test.to_parquet("input/test.parquet")

names = train.features["label"].names
pd.DataFrame({"label": range(len(names)), "intent": names}).to_csv("input/intents.csv", index=False)
```

```python
import pandas as pd

msgs = pd.read_parquet("data/simulated/messages")
clean = pd.read_parquet("data/processed/banking77_clean.parquet")

# Intents that never appear in the stream
used = set(msgs["text"].map(dict(zip(clean["text"], clean["intent"]))))
print(sorted(set(clean["intent"]) - used))

# Share of stream sentences that come from the test split
test_texts = set(clean.loc[clean["split"] == "test", "text"])
print(msgs["text"].isin(test_texts).mean())
```

```python
import pandas as pd

msgs = pd.read_parquet("data/simulated/messages")
clean = pd.read_parquet("data/processed/banking77_clean.parquet")

# 1. Missing intents: should be an empty list
used = set(msgs["text"].map(dict(zip(clean["text"], clean["intent"]))))
print(sorted(set(clean["intent"]) - used))

# 2. Share of test-split sentences: should be close to 0
test_texts = set(clean.loc[clean["split"] == "test", "text"])
print(msgs["text"].isin(test_texts).mean())

# 3. Daily fraud rate: should be roughly flat
report = pd.read_parquet("data/simulated/daily_report.parquet")
print(report[["day", "fraud_rate"]].to_string(index=False))
```
